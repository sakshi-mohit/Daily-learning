```python
""""
ATM system using oop( class and object)

"""

class ATM:
    def init_(self,name,balance):
        self.name = name 
        self.balance = balance

    def check_balance(self):
        print("Balance :", self.balance)

    def deposit(self, amount):
        self.balance += amount
        print("Deposited: ",amount)

    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance = amount
            print("Withdraw:", amount)
        else:
            print("Insufficient Balance")
        
```

```python

def main():
    user1 = ATM("sakshi",100000)
    user1.check_balance()
    user1.deposit(5000)
    user1.withdraw(3000)

if __name__ == "main" :
    main()
```

```python
"""
Poject :shopping CArt system using oop
Description: User can add/ remove / view items using class and object

"""
```

```python
class Shoppingcart:   # this is class

    def __init__(self):   # constructor (runs automatically)
        self.items = {}   # empty dictionary

    def add_item(self, name, price):
        self.items[name] = price
        print(f"{name} added successfully")

    def remove_item(self, name):
        if name in self.items:
            del self.items[name]
            print(f"{name} removed")
        else:
            print("Item not found")

    def view_cart(self):
        if not self.items:
            print("Cart is empty")
        else:
            print("Items in cart:")
            for name, price in self.items.items():
                print(f"{name}: {price}")

    def total_price(self):
        total = sum(self.items.values())
        print("Total Price:", total)
```

```python
def main():

    cart = Shoppingcart()

    while True:

        print("\n1. Add Item")
        print("2. Remove Item")
        print("3. View Cart")
        print("4. Total Price")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            name = input("Enter item name: ")
            price = float(input("Enter item price: "))   # convert to number
            cart.add_item(name, price)

        elif choice == "2":
            name = input("Enter item name to remove: ")
            cart.remove_item(name)

        elif choice == "3":
            cart.view_cart()

        elif choice == "4":
            cart.total_price()

        elif choice == "5":
            print("Thank you!")
            break

        else:
            print("Invalid choice")
        

if __name__ == "__main__":
    main()
```
