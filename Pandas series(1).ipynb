Series from lists

```python
# string 
import numpy as np
import pandas as pd
country = ["India","Pakistan","china","Iran","USA"]
pd.Series(country)
```

```python
#integers
runs = [13,24,56,78,100]
pd.Series(runs)
```

```python
# custom index
marks = [65,78,75,80]
subjects = ["java","DS","python","sql"]

pd.Series(marks,index = subjects)
```

```python
#setting a name

marks = (pd.Series(marks,index = subjects,name = "kim ke marks"))
print(marks)
```

Series from dict

```python
marks_series = {
    "maths" : 67,
    "English" :78,
    "Science" :89,
    "history" :88
}

marks_series = pd.Series(marks,name = "jungkook ke marks")
marks_series
```
