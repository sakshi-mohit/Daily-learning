#Working with mathematical formulas

```python
import numpy as np
a = np.arange(10)
np.sin(a)
```

```python
#sigmoid
def sigmoid(array):
    return 1/(1 + np.exp(-array))

a = np.arange(10)

sigmoid(a)
```

```python
#mean square error

actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
```

```python
def mse(actual,predicted):
    return np.mean((actual-predicted)**2)

mse(actual,predicted)    
```

```python
#plotting graph
# x = y
import matplotlib.pyplot as plt
x =np.linspace(-10,10,100)
y = x

plt.plot(x,y)
```

```python
# y = x^2
x = np.linspace(~10,10,100)
y = x**2
plt.plot(x,y)
```

```python
# y = sin(x)
x = np.linspace(~10,10,100)
y = np.sin(x)

plt.plot(x,y)
```

```python
# y = xlog(x)
x = np.linspace(-10,10,100)
y = x * np.log(x)
plt.plot(x,y)
```

```python
#sigmoid
x = np.linspace(-10,10,100)
y = 1/(1+np.exp(-x))
plt.plot(x,y)
```
