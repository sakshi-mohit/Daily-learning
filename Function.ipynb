# Let's create a function(with docstring)

```python
def is_even(num):
  """
  This function returns if a given number is odd or even
  input - any valid integer
  output - odd/even
  created on - 16th Nov 2022
  """
  if type(num) == int:
    if num % 2 == 0:
      return 'even'
    else:
      return 'odd'
  else:
    return 'pagal hai kya?'
```

```python
# function
# function_name(input)
for i in range(1,11):
    x = is_even(i)
    print(x)
    
    
```

```python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.

def multiply(*args):
    products = 1

    for i in args:
      products = products * i

    return products
```

```python
multiply(1,54,6)
```

```python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.

def multiply(*kwargs):
  product = 1

  for i in kwargs:
    product = product * i

  return product
  print (product)
```

# variable scope

```python
# Local scope 

def great():
    
    message = "hello sakshi"
    print(message)

great()

    
```

```python
# Global scope 

name = "sakshi_MOhit"

def show():
    print(name)

show()    
```

```python
# Enclosing scope

def outer():
    x = 10

    def inner():
        print(x)
    inner()    

outer()
        
```

```python
# LEBT

x = "Global"

def outer():
    x = "Enclosing"

    def inner():
        x = "Local"
        print(x)

    inner()
outer()    
    
```
