```python
# finally
# else

try:
    f = open("sample.txt","r")
except FileNotFoundError:
    print("file nhi mili")
except Exception:
    print("Kuch toh lafda hai")
else:
    print(f.read())
finally:
    print("Yeh toh print hoga hi")
    
```

```python
# Raise -> throw

class ATM():
    
    def __init__(self,balance):
        self.balance = balance
        
    def withdraw(self,amount):
        if amount < 0:
            raise Exception("__Amount can't be -ve__")
        if self.balance < amount:
            raise Exception("__Paise nhi hai tumhare pass__")

        self.balance = self.balance - amount    

obj = ATM(10000)
try:
    obj.withdraw(-5000)
except Exception as e:
    print(e)
else:
    print(obj.balance)
    
```
