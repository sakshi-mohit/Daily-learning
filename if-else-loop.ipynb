```python
# min of 3 no.
a = int(input("first no."))
b = int(input("second no."))
c = int(input("Third noo."))

if a < b and a < c:
    print("Smallest is ",a)
elif b < c:
    print("Smallest is ",b)
else:
    print("Smallest is ",c)
```

```python
#menu driven calculator

fnum = int(input("Enter first no."))
snum = int(input("Enter sec no."))

op = input("Enter any operator (+,-,*,/): ")

if op == "+" :
    print(fnum + snum)
elif op == "-":
     print(fnum - snum)
elif op == "*":
     print(fnum * snum)
else:
     print(fnum / snum)
```

```python
# guessing game
import random
jackpot = random.randint(1,100)

guess = int(input("Guess karo : "))

counter = 1
while guess != jackpot:
    if guess < jackpot:
        print("galat ! higher")
    else:
        print("galat ! lower")
    guess = int(input("Guess karo : "))  
    counter += 1

else:
    print("Correct guess")
    print("attempts: ",counter)
```

```python
""" The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year.
You have to write a program to find out the population at the end of each of the last 10 years."""

population = 10000

for year in range (10,0,-1):
    population = population * 1.10
    print(year,population)
```
