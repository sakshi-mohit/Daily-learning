# data processing using tuple ,set, Dictionary

# 1.use tuple to  handle fixed user details 
# 2.Use set to remove duplicate users
# 3. use dictionary to analyze frequency of users




```python
# convert  list into tuple 
# access elements using indexing
# print 1st and last user
```

```python
users = ["sakshi","atharv","tanmay","sakshi", "bhavik"]

users_tuple = tuple(users)

print("tuple data:", users_tuple)
print("first user:",users_tuple[0])
print("Last user:",users_tuple[-1])
```

```python
#  convert data into set 
# print unique users
```

```python
users = ["sakshi","atharv","tanmay","sakshi", "bhavik"]

users_set = set(users)
print("unique users :",users_set)
```

```python
# Count occurrences of each user.
# Store result in dictionary.

users = ["sakshi","atharv","tanmay","sakshi", "bhavik"]

frequency = {}

for user in users:
    frequency[user] = frequency.get(user , 0) + 1

print(frequency)    
```
