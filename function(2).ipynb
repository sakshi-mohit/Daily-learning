```python
# lambda function
a = lambda x,y:x+y
a(5,2)
```

```python
# check if a string has "a"

a = lambda s:"a" in s
a("hello")
```

```python
# odd and even 
a = lambda x: "even" if x % 2 == 0 else "odd"
a(6)
```

```python
# Higher Order function

# example

def square(x):
    return x**2

def transform(f,L):
    output = []
    for i in  L:
        output.append(f(i))

    print(output)    

L = [1,2,22,40]
transform(square, L)    
```

```python
# Map

list(map(lambda x : x**2 , [1,2,3,4]))
```

```python
# even and odd labelling of list items

L = [1,2,3,4,5]
list(map(lambda x:'even' if x%2 == 0 else 'odd' , L))
```

```python
# numbers grater than 5

L = [3,2,5,6,9]
list(filter(lambda x: x>5 , L))
```

```python
fruits = ["apple ", "guava", "mango" ]
list(filter(lambda x: x.startswith("a"),fruits))
```

```python
# sum of all list

import functools

functools.reduce(lambda x,y: x+y, [1,2,3,4,4,4,])
```

```python
# find min

functools.reduce(lambda x,y :x if x < y else y , [23,56,94])
```
